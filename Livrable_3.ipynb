# Optimisation des tournées de livraison (VRP)

## Contexte

CesiCDP doit réaliser une étude afin d'optimiser les tournées de livraison de telle sorte à :

- Relier un sous-ensemble de clients
- Revenir au point de départ
- Minimiser la durée totale de la tournée

## Reformulation formelle du contexte

Soit un graphe $G = (V, R)$ où :

- $V = \{1, \dots, n\}$ représente les clients, qui sont les **sommets**.
- $R = \{(i, j) \mid i, j \in V,\ i \neq j\}$ représente l'ensemble des routes reliant les villes, qui sont les **arêtes**.

### Paramètres du graphe

| Paramètre | Description |
|---|---|
| $C_{ij}$ | Coût (ou durée) du trajet de $i$ vers $j$ |
| $D_i$ | Demande (quantité à livrer) pour le client $i$. Pour le dépôt : $D_0 = 0$ |
| $K$ | Nombre de camions disponibles (ou ensemble des camions) |
| $Q$ | Capacité maximale de chaque camion (la somme des demandes sur une route ne doit pas dépasser $Q$) |

## Formalisation du problème

### Problème de décision associé

- **Données** : un graphe $G$
- **Question** : existe-t-il un circuit passant une fois par chaque ville ?

### Problème d'optimisation

- **Données** : un graphe $G$
- **Question** : existe-t-il un circuit passant une fois par chaque ville $V$ avec un coût $C$ minimal ?

## Complexité théorique

### Réduction au problème du voyageur de commerce

Le problème **VRP** (*Vehicle Routing Problem*) est une généralisation du problème **TSP** (*Traveling Salesperson Problem*).

En posant $K = 1$ (un seul camion) et une capacité maximale $Q = \infty$, le problème revient simplement à trouver le chemin le plus court passant **une seule fois par chaque ville** et **revenant au dépôt**.
