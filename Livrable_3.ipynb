# Optimisation des tournées de livraison (VRP)

## Contexte

CesiCDP doit réaliser une étude afin d'optimiser les tournées de livraison de telle sorte à :

- Relier un sous-ensemble de clients
- Revenir au point de départ
- Minimiser la durée totale de la tournée

## Reformulation formelle du contexte

Soit un graphe $G = (V, R)$ où :

- $V = \{1, \dots, n\}$ représente les clients, qui sont les **sommets**.
- $R = \{(i, j) \mid i, j \in V,\ i \neq j\}$ représente l'ensemble des routes reliant les villes, qui sont les **arêtes**.

### Paramètres du graphe

| Paramètre | Description |
|---|---|
| $C_{ij}$ | Coût (ou durée) du trajet de $i$ vers $j$ |
| $D_i$ | Demande (quantité à livrer) pour le client $i$. Pour le dépôt : $D_0 = 0$ |
| $K$ | Nombre de camions disponibles (ou ensemble des camions) |
| $Q$ | Capacité maximale de chaque camion (la somme des demandes sur une route ne doit pas dépasser $Q$) |

## Formalisation du problème

### Problème de décision associé

- **Données** : un graphe $G$
- **Question** : existe-t-il un circuit passant une fois par chaque ville ?

### Problème d'optimisation

- **Données** : un graphe $G$
- **Question** : existe-t-il un circuit passant une fois par chaque ville $V$ avec un coût $C$ minimal ?

## Complexité théorique

### Réduction au problème du voyageur de commerce

Le problème **VRP** (*Vehicle Routing Problem*) est une généralisation du problème **TSP** (*Traveling Salesperson Problem*).

En posant $K = 1$ (un seul camion) et une capacité maximale $Q = \infty$, le problème revient simplement à trouver le chemin le plus court passant **une seule fois par chaque ville** et **revenant au dépôt**.

### Programme Linéaire en nombres Entiers (PLNE)

Le PLNE (Programmation Linéaire en Nombres Entiers) est une méthode de modélisation et de résolution de problèmes d’optimisation. Elle consiste à définir des variables de décision entières, une fonction objectif linéaire à minimiser ou maximiser, et des contraintes linéaires que les variables doivent respecter.

- **Les variables de décisions**

$x_{ij}^k$=1: Si un camion a enprunter une route reliant deux villes. <br>
$x_{ji}^k$=0: Sinon.

- **La fonction objectif**

On veut minimiser le cout total du transport, c'est-à-dire la somme des coûts de tous les trajets effectués par tous les camions.

$$\min Z = \sum_{k \in K} \sum_{i \in V} \sum_{j \in V} c_{ij} \cdot x_{ij}^k$$

| Somme | Parcourt | Rôle |
|---|---|---|
| $\sum_{k \in K}$ | tous les camions | on cumule le coût de chaque camion |
| $\sum_{i \in V}$ | tous les villes de départ | on considère chaque origine possible |
| $\sum_{j \in V}$ | tous les villes d'arrivée | on considère chaque destination possible |

- **Contraintes**

| # | Contrainte | Formule | Rôle |
|---|---|---|---|
| 1 | Visite unique | $\sum_{k \in K} \sum_{j \in V} x_{ij}^k = 1 \quad \forall i \in \{1, \dots, n\}$ | Chaque client est visité exactement une fois, par un seul camion. |
| 2 | Conservation du flux | $\sum_{i \in V} x_{ij}^k = \sum_{p \in V} x_{jp}^k \quad \forall j \in \{1, \dots, n\}, \forall k \in K$ | Si un camion arrive chez un client, il en repart : les tournées sont continues. |
| 3 | Départ du dépôt | $\sum_{j \in V} x_{0j}^k \le 1 \quad \forall k \in K$ | Chaque camion part du dépôt au plus une fois (il peut rester inutilisé). |
| 4 | Retour au dépôt | $\sum_{i \in V} x_{i0}^k \le 1 \quad \forall k \in K$ | Chaque camion utilisé revient au dépôt une seule fois. |
| 5 | Capacité | $\sum_{i \in V} \sum_{j \in V} d_j \cdot x_{ij}^k \le Q \quad \forall k \in K$ | La somme des demandes livrées par un camion ne dépasse pas sa capacité $Q$. |
| 6 | Élimination des sous-tours (MTZ) | $u_i^k - u_j^k + Q \cdot x_{ij}^k \le Q - d_j \quad \forall i \neq j \in \{1, \dots, n\}, \forall k \in K$<br>et $d_i \le u_i^k \le Q$ | Interdit les boucles qui ne passent pas par le dépôt. |
| 7 | Nature des variables | $x_{ij}^k \in \{0, 1\}$ | Les variables sont binaires : décision oui/non (c'est le « NE » du PLNE). |
